## 🧾 **Conclusão**

### 🔸 **Performance de Produtos**

**Produtos mais lucrativos**

Apple Music lidera em receita total (R$ 125,4M), seguido de iMac 27-inch (R$ 124M) e iPad mini 5ª Geração (R$ 123,9M). Os volumes de unidades vendidas são bastante próximos entre os top 10 (entre 63 e 65 mil unidades), o que indica que a diferença de receita é explicada principalmente pelo preço unitário de cada produto.

**Distribuição por categoria**

Accessories é a categoria com maior volume de vendas (~162 mil transações), seguida de Smartphone (~151 mil) e Audio (~128 mil). Smart Speaker e Streaming Device registraram os menores volumes, provavelmente por representarem um portfólio menor de produtos.

**Produtos recém-lançados**

O MacBook Pro 14-inch (lançado em maio/2024) lidera entre os lançamentos mais recentes com 1.156 vendas, seguido de Magic Trackpad (1.109) e MacBook Pro Touch Bar (970). Produtos lançados há mais tempo dentro do período naturalmente acumulam mais vendas. O Apple TV+, lançado em novembro/2024, ainda está no início da curva de adoção com apenas 75 registros.

**Correlação preço × volume**

A correlação de Pearson entre preço e volume vendido foi de -0,02, indicando ausência de relação linear entre as variáveis. Produtos mais caros não vendem necessariamente menos, o que é coerente com o comportamento típico dos consumidores Apple.

### 🔸 **Desempenho Geográfico e por Unidade**

**Faturamento por loja**

Apple Chadstone (Melbourne) lidera com R$ 165,2M em receita e 27.851 vendas. As 6 primeiras lojas do ranking têm desempenho bastante similar, com faturamento entre R$ 161M e R$ 165M. A partir da 7ª posição (Apple Dubai Mall), a receita cai pela metade (~R$ 84M), sugerindo dois grupos distintos de desempenho entre as unidades. 

**Liderança geográfica**

Os Estados Unidos lideram com R$ 1,23 bilhão em receita e 207.728 vendas — mais que o dobro do segundo colocado. Austrália (R$ 577M) e China (R$ 575M) vêm logo atrás com desempenhos muito próximos entre si. México e Alemanha fecham o top 10 com volumes menores, indicando mercados ainda em desenvolvimento para a marca.
